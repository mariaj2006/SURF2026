```python
import numpy as np
path = "/home/caltech-msanche3/data/jades_eazy_output/fit_data/jades_run_60k_183348.zout.fits"

# this data maintained the fixed redshifts provided by the jades catalog, we only fit for other properties
jades_60k = Table.read(path)


# Assuming your catalog is a NumPy array or Pandas DataFrame
z_mask = (jades_60k['z_phot'] >= 13) & (jades_60k['z_phot'] < 15)
bin_galaxies = jades_60k[z_mask]
# 211th object is id=183348, our target galaxy

# Sort descending so the absolute heaviest galaxy is at index 0
sorted_indices = np.argsort(bin_galaxies['mass'])[::-1]
sorted_galaxies = bin_galaxies[sorted_indices]
mask = sorted_galaxies['mass'] != 0
sorted_galaxies = sorted_galaxies[mask]
# item 41 is id=183348
print(len(sorted_galaxies['id','mass']))
```

```python
cumul_mass_limits = []
cumul_mass_lo = []
cumul_mass_hi = []

print('tot:',sum(sorted_galaxies['mass']))
print('tot lo:',sum(sorted_galaxies['mass_p'][:,1]))
print('tot hi:',sum(sorted_galaxies['mass_p'][:,3]))

# loop through some binned sums
steps = [93469042286.59503, 9000000, 7000000, 1000]

for i in range(4):
    running_mass_sum = 0
    mass_lo_sum = 0
    mass_hi_sum = 0
    mask = sorted_galaxies['mass'] >= steps[i]
    cumalative_gal = sorted_galaxies[mask]
    print(f'num galaxies at mass >= {steps[i]}:',len(cumalative_gal))
    for galaxy in cumalative_gal:
        running_mass_sum += galaxy['mass']
        mass_lo_sum += galaxy['mass_p'][1]
        mass_hi_sum += galaxy['mass_p'][3]
    cumul_mass_limits.append(running_mass_sum)
    cumul_mass_lo.append(mass_lo_sum)
    cumul_mass_hi.append(mass_hi_sum)
print('limit:',cumul_mass_limits)
print('lo:',cumul_mass_lo)
print('hi:',cumul_mass_hi)
```
